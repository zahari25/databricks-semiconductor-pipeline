```python
df_bronze = spark.table("devsquad_bt.semiconductor_project.bronze_signal_data")
print("Columnas totales:", len(df_bronze.columns))
print("Registros:", df_bronze.count())
```

```python
from pyspark.sql.functions import col, sum as _sum

null_counts = df_bronze.select([
    _sum(col(c).isNull().cast("int")).alias(c) for c in df_bronze.columns
])

# Monstramos solo las columnas que tienen nulos 
nulls_dict = null_counts.collect()[0].asDict()
columnas_con_nulos = {k: v for k, v in nulls_dict.items() if v > 0}

print(f"columnas que tienen valores nulos: {len(columnas_con_nulos)}")
print("Ejemplo de las primeras 10 columnas con nulos:")
for i, (col_name, nulls) in enumerate(list(columnas_con_nulos.items())[:10]):
    print(f"{col_name}: {nulls} nulos")
```

```python
from pyspark.sql.functions import col, sum as _sum

df_silver = spark.table("devsquad_bt.semiconductor_project.silver_signal_data")

# Contamos nulos por columna
total_rows = df_silver.count()
null_counts = df_silver.select([
    _sum(col(c).isNull().cast("int")).alias(c) for c in df_silver.columns
]).collect()[0].asDict()

# Elegimos solo las columnas que tienen menos del 50% de nulos
columnas_buenas = [c for c, nulos in null_counts.items() if (nulos / total_rows) < 0.50]

print(f"Columnas originales: {len(df_silver.columns)}")
print(f"Columnas que se quedan (menos del 50% nulos): {len(columnas_buenas)}")

# Creamos la version limpia
df_silver_limpia = df_silver.select(columnas_buenas)

# Guardamos la version limpia (sobrescribimos Silver)
df_silver_limpia.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(
    "devsquad_bt.semiconductor_project.silver_signal_data"
)

print("Tabla Silver actualizada (version limpia)")
```

# Capa Gold
#### Creacion de dos tablas utiles para el analisis:
##### 1. Resumen de nulos -> para ver la calidad de los datos
##### 2. Estadisticas basicas de algunos sensores

```python
from pyspark.sql.functions import col, sum as _sum, lit

df = spark.table("devsquad_bt.semiconductor_project.silver_signal_data")
total_rows = df.count()
null_counts = df.select([
    _sum(col(c).isNull().cast("int")).alias(c) for c in df.columns
]).collect()[0].asDict()

#Convertimos a DataFrame
data = [(col_name, nulls, round(nulls/total_rows*100, 2))
        for col_name, nulls in null_counts.items()]

df_nulls = spark.createDataFrame(data, ["columna", "nulos", "porcentaje_nulos"])

df_nulls.write.mode("overwrite").saveAsTable(
    "devsquad_bt.semiconductor_project.gold_nulls_summary"
)

print("Tabla Gold de nulos creada")
display(df_nulls.orderBy("nulos", ascending=False).limit(10))
```

```python
from pyspark.sql.functions import col, mean, min, max, stddev, sum as _sum

df = spark.table("devsquad_bt.semiconductor_project.silver_signal_data")

# Seleccionamos solo las columnas de sensores (excluimos Time)
sensor_cols = [c for c in df.columns if c.startswith("Sensor_")]

print(f"Se van a procesar {len(sensor_cols)} sensores...")

# Creamos las estadisticas
stats = []

for c in sensor_cols:
    row = df.agg(
        mean(col(c)).alias("promedio"),
        min(col(c)).alias("minimo"),
        max(col(c)).alias("maximo"),
        stddev(col(c)).alias("desviacion"),
        _sum(col(c).isNull().cast("int")).alias("nulos")
    ).collect()[0]

    stats.append((
        c, 
        float(row["promedio"]) if row["promedio"] is not None else None, 
        float(row["minimo"]) if row["minimo"] is not None else None, 
        float(row["maximo"]) if row["maximo"] is not None else None, 
        float(row["desviacion"]) if row["desviacion"] is not None else None, 
        int(row["nulos"])
    ))

# Convertimos a Dataframe
df_stats = spark.createDataFrame(
    stats,
    ["sensor", "promedio", "minimo", "maximo", "desviacion", "nulos"]
)

# Guardamos como tabla Gold
df_stats.write.mode("overwrite").saveAsTable(
    "devsquad_bt.semiconductor_project.gold_sensor_stats"
)

print("Tabla Gold de estadistics creada correctament")
display(df_stats.orderBy("nulos", ascending=False).limit(10))
```
